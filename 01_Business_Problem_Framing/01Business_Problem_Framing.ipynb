# Business_Problem_Framing
This Project deals with the ___Identification___ of any existent/prevailing Business Problem. The actual purpose of this project is to ***learn and think like a data-scientist***, who not just casually looks at the problem and suggest the ways to tackle, but deeply learns by sticking to Business Objective and find pathways of creating insightful results by critically examining the factors responsible for the state of the Business, these include:
* Briefly understanding the Business-Objective, helpful in tackling the prevailing Business-Problem.
* Learning about the problem by analyzing the kay factors responsible.
* Linking and classifying the hidden/unseen patterns causing the Business-Problem.
* Developing a calculative structure/model which based on the past data... and generates successful business results.    
### Identifying a Real-Business Problem:
In this project, I will be taking a `demand-sales` problem structure concerned with the sales and management of a ## Retail Company whose Business is set and based on marketing through Physical Grocery Stores and Online Platforms.
    
The Management noticed several problems regarding their sales through the stores:
* Some products frequently go out of stock.
* Some products remain unsold for long periods.
* Demand varies considerably between products & locations.
* Promotions affect Sales.
* Weekends & Holidays affects Demand.        

Based on the listed points above, the `Business-Problem` can be described as: 
___"How can we better understand and predict product demand so that we can maintain the right inventory?"___

### Rightful DS Problem Statement:
**Learning about the Business-Problem itself does not solves it**, rather it creates a base for actually understanding the objectives to successfully manage and certainly eradicate a caused intervention in the ongoing business!
Here comes the role of a Data-Scientist, this person then understands the Business-Objective and filter out the exact points and loopholes that are causing the struggle of a business...

The `Data-Scientist` targets the responsible or key areas of any business struggle by having thoughtful insights over the past & previous reports of the corporation/business. 
In building a promising summary of the listed factors, the person must go through the series of tasks which are usually based on the kind of Business-Problem that person is actually dealing with.

`For my chosen Business-Problem(demand-inventory management)`, the advised Data-Science Problem Statement is: ___"Analyze historical sales, product, store, seasonal, promotional, and pricing data to identify demand patterns and estimate future product demand."___

In short, a particular business-problem can have multiple DS Problem Statements... which depends on the target the Business is tracking!!!

## Business Problem - Retail Demand & Inventory Optimization
The company struggles to maintain optimal inventory because product demand varies across time, products, locations, seasons, and promotional periods.

## Business objective:
Improve inventory planning and reduce both stockouts and excess inventory by understanding and eventually predicting product demand.

## Data science problem:  
Identify the factors and patterns influencing product demand and develop a data-driven approach for estimating future demand.
    
## A Potential ML Problem:
`Time-series forecasting / regression`. The Model predicts or generates the future sales-demand structural anaylsis for a better and effecient inventory management.

## Required data:
* Historical sales
* Product information
* Store/location
* Date/time
* Price
* Promotions
* Holidays
* Inventory levels, if available
* Weather, if relevant    

The evident dataset used for this Business-Problem is being taken from kaggle, namely: `Retail Store Inventory Forecasting Dataset`

## Expected output:
Forecasted demand for a product/store over a future period. Suppose, the regression fits well on our data and actually let the business(retail store) predict the scale of sales as to demand in a locality...
Imagine how effeciently the business can use this model to set the ideals of a successful business.
    These predictive models can expand their scope in multiple domains like extending the priorties of a business-sales and market value by knowing about their competing and rival businesses, also managing the portfolio reports of their employees and many more... 

## Business decision:
Determine how much inventory should be stocked and when. This actually shapes the business ideals and growth in a dramatic way never thought before.
