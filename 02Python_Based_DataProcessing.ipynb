### Data Building using Python
How can raw retail inventory records be processed and transformed into meaningful store-, product-, and inventory-level metrics that can support demand and inventory decisions?...

Loading the subject-dataset and reading it through pandas library. The python program reads the actual raw data and presents in a tabular format consisting of data-entry rows into their entitled columns!

```python
subject_dataset["Date"] = pd.to_datetime(subject_dataset['Date'])
subject_dataset['Date'].dtype
```

```python
subject_dataset['Year'] = subject_dataset['Date'].dt.year
subject_dataset['Month'] = subject_dataset['Date'].dt.month
subject_dataset['Day'] = subject_dataset['Date'].dt.day
```

```python
import numpy as np

subject_dataset['Inventory Level'] = pd.to_numeric(subject_dataset['Inventory Level'], errors='coerce')
conditions = [subject_dataset['Inventory Level'] <= 100,
              subject_dataset['Inventory Level'] <= 250 ]
choices = ['Low', 'Medium']
subject_dataset['Inventory Status'] = np.select(conditions, choices, default='Healthy')
```

```python
subject_dataset['Units Sold'] = pd.to_numeric(subject_dataset['Units Sold'], errors='coerce')

subject_dataset['Inventory Utilization'] = np.where(
    subject_dataset['Inventory Level'] > 0,
    (subject_dataset['Units Sold'] / subject_dataset['Inventory Level']) * 100,
    0.0
)
```

```python
subject_dataset["Order Difference"] = (
    subject_dataset["Units Ordered"] - subject_dataset["Demand Forecast"])
subject_dataset["Sales Difference"] = (
    subject_dataset["Units Sold"] - subject_dataset["Demand Forecast"])
```

```python
subject_dataset.head()
```

```python
for i in range(len(subject_dataset)):
    difference = subject_dataset.loc[i, "Sales Difference"]
    if difference > 0:
        subject_dataset.loc[i, "Demand Status"] = "Higher than Forecast"
    elif difference < 0:
        subject_dataset.loc[i, "Demand Status"] = "Lower than Forecast"
    else:
        subject_dataset.loc[i, "Demand Status"] = "Same as Forecast"
```

```python
subject_dataset[['Order Difference', 'Sales Difference', 'Demand Status']].head()
```

```python
store_summary = subject_dataset.groupby("Store ID").agg(
    Total_Units_Sold=("Units Sold", "sum"),
    Total_Units_Ordered=("Units Ordered", "sum"),
    Average_Inventory=("Inventory Level", "mean"),
    Average_Demand_Forecast=("Demand Forecast", "mean")
).reset_index()
```

```python
store_summary.head()
```

```python
product_summary = subject_dataset.groupby("Product ID").agg(
    Total_Units_Sold=("Units Sold", "sum"),
    Total_Units_Ordered=("Units Ordered", "sum"),
    Average_Inventory=("Inventory Level", "mean"),
    Average_Demand_Forecast=("Demand Forecast", "mean")
).reset_index()
```

```python
product_summary.head()
```

```python
low_inventory = subject_dataset[subject_dataset["Inventory Status"] == "Low"]
high_demand_products = product_summary.sort_values(
    by="Total_Units_Sold",
    ascending=False
)
```
