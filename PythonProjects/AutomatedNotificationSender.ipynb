```python
import json
import os
import time
from datetime import datetime
from plyer import notification
fname = "notifications.json"

def load_notifications():
    if os.path.exists(fname):
        if os.path.getsize(fname) == 0: 
            return []
        try:
            with open(fname, "r") as f:
                return json.load(f)
        except json.JSONDecodeError:
            print("Warning: JSON file is corrupted or empty. Starting fresh.")
            return []
        return[]    
        f = open(fname, "r")
        notifications = json.load(f)
        f.close()
        return notifications
    else:
        return []
def save_notifications(notifications):
    f = open(fname, "w")
    json.dump(notifications, f, indent=4)
    f.close()
def send_notification(title, message):
    notification.notify(
        title=title,
        message=message,
        app_name="Python Notification Sender",
        timeout=10
    )
def add_notification(notifications):
    print("\n---------- ADD NOTIFICATION ----------")

    title = input("Enter notification title: ")
    message = input("Enter notification message: ")

    remind_time = input("Enter time (HH:MM): ")

    notification_data = {
        "title": title,
        "message": message,
        "time": remind_time,
        "sent": False
    }

    notifications.append(notification_data)

    save_notifications(notifications)

    print("Notification added successfully!")
def view_notifications(notifications):
    print("\n---------- SAVED NOTIFICATIONS ----------")

    if len(notifications) == 0:
        print("No notifications found.")
    else:
        i = 1

        for item in notifications:
            print("\nNotification", i)
            print("Title:", item["title"])
            print("Message:", item["message"])
            print("Time:", item["time"])
            print("Sent:", item["sent"])

            i = i + 1

    print("------------------------------------------")
def send_now():
    print("\n---------- SEND NOTIFICATION ----------")

    title = input("Enter notification title: ")
    message = input("Enter notification message: ")

    send_notification(title, message)

    print("Notification sent!")
def check_notifications(notifications):

    current_time = datetime.now().strftime("%H:%M")

    for item in notifications:

        if item["time"] == current_time and item["sent"] == False:

            send_notification(
                item["title"],
                item["message"]
            )
            item["sent"] = True
            save_notifications(notifications)
            print("\nNotification sent:", item["title"])
notifications = load_notifications()

while True:

    print("\n========================================")
    print("     AUTOMATION NOTIFICATION SENDER")
    print("========================================")
    print("1. Send notification now")
    print("2. Add scheduled notification")
    print("3. View notifications")
    print("4. Start notification checker")
    print("5. Exit")
    print("========================================")

    choice = input("Enter your choice: ")

    if choice == "1":
        send_now()

    elif choice == "2":
        add_notification(notifications)

    elif choice == "3":
        view_notifications(notifications)

    elif choice == "4":
        print("\nNotification checker started.")
        print("Keep this program running.")
        print("Press CTRL + C to stop it.")

        while True:
            check_notifications(notifications)
            time.sleep(20)

    elif choice == "5":
        save_notifications(notifications)
        print("\nNotifications saved. Goodbye!")
        break

    else:
        print("Invalid choice. Please try again.")
```
