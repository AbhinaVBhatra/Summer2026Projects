```python
##BankAccountSystem---------------------

import json
import os
fname = "bank_accounts.json"

def load_accounts():
    if os.path.exists(fname):
        f = open(fname, "r")
        accounts = json.load(f)
        f.close()
        return accounts
    else:
        return []

def save_accounts(accounts):
    f = open(fname, "w")
    json.dump(accounts, f, indent=4)
    f.close()

def create_account(accounts):
    print("\n---------- CREATE ACCOUNT ----------")

    name = input("Enter account holder name: ")
    account_no = input("Enter account number: ")
    balance = float(input("Enter initial deposit: "))

    account = {
        "name": name,
        "account_no": account_no,
        "balance": balance
    }

    accounts.append(account)

    save_accounts(accounts)

    print("Account created successfully!")
    
def view_accounts(accounts):
    print("\n---------- ALL ACCOUNTS ----------")

    if len(accounts) == 0:
        print("No accounts found.")
    else:
        i = 1

        for account in accounts:
            print("\nAccount", i)
            print("Name:", account["name"])
            print("Account Number:", account["account_no"])
            print("Balance:", account["balance"])

            i = i + 1

    print("----------------------------------")

def search_account(accounts):
    print("\n---------- SEARCH ACCOUNT ----------")

    account_no = input("Enter account number: ")

    found = False

    for account in accounts:
        if account["account_no"] == account_no:

            print("\nAccount found!")
            print("Name:", account["name"])
            print("Account Number:", account["account_no"])
            print("Balance:", account["balance"])

            found = True
            break

    if found == False:
        print("Account not found.")

def deposit_money(accounts):
    print("\n---------- DEPOSIT MONEY ----------")

    account_no = input("Enter account number: ")

    found = False

    for account in accounts:
        if account["account_no"] == account_no:

            amount = float(input("Enter amount to deposit: "))

            if amount > 0:
                account["balance"] = account["balance"] + amount

                save_accounts(accounts)

                print("Money deposited successfully!")
                print("New balance:", account["balance"])
            else:
                print("Amount should be greater than zero.")

            found = True
            break

    if found == False:
        print("Account not found.")

def withdraw_money(accounts):
    print("\n---------- WITHDRAW MONEY ----------")

    account_no = input("Enter account number: ")

    found = False

    for account in accounts:
        if account["account_no"] == account_no:

            amount = float(input("Enter amount to withdraw: "))

            if amount <= 0:
                print("Amount should be greater than zero.")

            elif amount > account["balance"]:
                print("Insufficient balance.")

            else:
                account["balance"] = account["balance"] - amount

                save_accounts(accounts)

                print("Money withdrawn successfully!")
                print("Remaining balance:", account["balance"])

            found = True
            break

    if found == False:
        print("Account not found.")

def delete_account(accounts):
    print("\n---------- DELETE ACCOUNT ----------")

    account_no = input("Enter account number: ")

    found = False

    for account in accounts:
        if account["account_no"] == account_no:

            accounts.remove(account)

            save_accounts(accounts)

            print("Account deleted successfully!")

            found = True
            break

    if found == False:
        print("Account not found.")
accounts = load_accounts()

while True:

    print("\n================================")
    print("       BANK ACCOUNT SYSTEM")
    print("================================")
    print("1. Create account")
    print("2. View accounts")
    print("3. Search account")
    print("4. Deposit money")
    print("5. Withdraw money")
    print("6. Delete account")
    print("7. Exit")
    print("================================")

    choice = input("Enter your choice: ")

    if choice == "1":
        create_account(accounts)

    elif choice == "2":
        view_accounts(accounts)

    elif choice == "3":
        search_account(accounts)

    elif choice == "4":
        deposit_money(accounts)

    elif choice == "5":
        withdraw_money(accounts)

    elif choice == "6":
        delete_account(accounts)

    elif choice == "7":
        save_accounts(accounts)
        print("\nAccount records saved. Goodbye!")
        break

    else:
        print("Invalid choice. Please try again.")
```
