```python
import os
import shutil

def create_folders(folder):
    image_folder = os.path.join(folder, "Images")
    document_folder = os.path.join(folder, "Documents")
    video_folder = os.path.join(folder, "Videos")
    audio_folder = os.path.join(folder, "Audio")
    other_folder = os.path.join(folder, "Others")

    if not os.path.exists(image_folder):
        os.mkdir(image_folder)

    if not os.path.exists(document_folder):
        os.mkdir(document_folder)

    if not os.path.exists(video_folder):
        os.mkdir(video_folder)

    if not os.path.exists(audio_folder):
        os.mkdir(audio_folder)

    if not os.path.exists(other_folder):
        os.mkdir(other_folder)

def organize_files(folder):

    image_extensions = [".jpg", ".jpeg", ".png", ".gif", ".webp"]
    document_extensions = [".pdf", ".doc", ".docx", ".txt", ".xlsx", ".csv"]
    video_extensions = [".mp4", ".mkv", ".avi", ".mov", ".webm"]
    audio_extensions = [".mp3", ".wav", ".aac", ".m4a", ".flac"]

    files = os.listdir(folder)

    for file in files:
        file_path = os.path.join(folder, file)

        if os.path.isfile(file_path):
            extension = os.path.splitext(file)[1].lower()
            if extension in image_extensions:
                destination = os.path.join(folder, "Images", file)

            elif extension in document_extensions:
                destination = os.path.join(folder, "Documents", file)

            elif extension in video_extensions:
                destination = os.path.join(folder, "Videos", file)

            elif extension in audio_extensions:
                destination = os.path.join(folder, "Audio", file)

            else:
                destination = os.path.join(folder, "Others", file)

            shutil.move(file_path, destination)

            print("Moved:", file)

def show_files(folder):

    print("\n---------- FILES ----------")
    files = os.listdir(folder)
    for file in files:
        print(file)
    print("---------------------------")


print("================================")
print("       BULK FILE ORGANIZER")
print("================================")

folder = input("Enter the folder path: ")

if os.path.exists(folder):
    create_folders(folder)
    print("\nFiles before organizing:")
    show_files(folder)

    print("\nOrganizing files...\n")
    organize_files(folder)
    print("\nFiles organized successfully!")
else:
    print("The folder does not exist.")
```
