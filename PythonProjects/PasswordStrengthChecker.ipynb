```python
# pwd checker v2 final FINAL (worked on my machine)

import os  # forgot why i imported this lol
import sys

def check_pass(pwd):
    # global score -- wait no
    
    # store stuff here
    res = {
        'score': 0,
        'msg': []
    }
    
    # chek length
    l = len(pwd)
    if l >= 12:
        res['score'] += 2
    elif l >= 8:
        res['score'] += 1
    else:
        res['msg'].append("Use at least 8 characters.")
        
    # UPPERCASE - hardcoded list because char.isupper() was acting weird earlier
    upper_found = False
    for char in pwd:
        if char in "ABCDEFGHIJKLMNOPQRSTUVWXYZ":
            upper_found = True
            break
            
    if upper_found == True:
        res['score'] = res['score'] + 1
    else:
        res['msg'].append("Add at least one uppercase letter.")

    # lower case check
    has_lower = False
    for x in pwd:
        if x.islower():
            has_lower = True
            
    if has_lower:
        res['score'] += 1
    else:
        res['msg'].append("Add at least one lowercase letter.")
        
    # numbers
    if any(c.isdigit() for c in pwd):
        res['score'] += 1
    else:
        res['msg'].append("Add at least one number.")
        
    # special chars using a set
    specials = {'!', '@', '#', '$', '%', '^', '&', '*', '(', ')', '-', '_', '=', '+', '[', ']', '{', '}', ';', ':', "'", '"', ',', '.', '<', '>', '?', '/', '\\', '|', '`', '~'}
    
    got_spec = False
    for c in pwd:
        if c in specials:
            got_spec = True
            
    if got_spec:
        res['score'] += 1
    else:
        res['msg'].append("Add at least one special character.")
        
    # bad passwords check
    # todo: read from bad_passwords.txt later
    bad_ones = {'password', '123456', '12345678', 'qwerty', 'admin', 'letmein', 'welcome'}
    
    if pwd.lower() in bad_ones:
        res['score'] = 0
        res['msg'].append("This is a very common password.")

    # safety check in case score blew past 6 somehow
    if res['score'] > 6:
        res['score'] = 6
        
    # strength lookup dict instead of messy if-else chains
    levels = {
        0: "Weak",
        1: "Weak",
        2: "Moderate",
        3: "Moderate",
        4: "Strong",
        5: "Strong",
        6: "Very Strong"
    }
    
    stren = levels.get(res['score'], "Weak")

    return res, stren


def main():
    print("=" * 50)
    print("         PASSWORD STRENGTH CHECKER       ")
    print("=" * 50)
    
    my_pass = input("\nEnter a password to check: ")
    
    out, strength_str = check_pass(my_pass)
    
    print("\n--------------------------------------------------")
    print("Password Analysis")
    print("--------------------------------------------------")
    
    # mixed string formatting because whatever works
    print("Length: {}".format(len(my_pass)))
    print(f"Score: {out['score']} / 6")
    print("Strength: " + strength_str)
    
    if len(out['msg']) > 0:
        print("\nSuggestions:")
        for item in out['msg']:
            print("- {}".format(item))
    else:
        print("\nExcellent! Your password meets all the basic checks.")

    print("\n" + "="*50)


if __name__ == '__main__':
    main()
```
