```python
#import os
#import sys

print("=" * 50)
print("         PASSWORD STRENGTH CHECKER       ")
print("=" * 50)

pwd = input("\nEnter a password to check: ")

score = 0
msg = []

l = len(pwd)
if l >= 12:
    score = score + 2
elif l >= 8:
    score = score + 1
else:
    msg.append("Use at least 8 characters.")

upper_found = False
for char in pwd:
    if char in "ABCDEFGHIJKLMNOPQRSTUVWXYZ":
        upper_found = True

if upper_found == True:
    score = score + 1
else:
    msg.append("Add at least one uppercase letter.")

has_lower = False
for x in pwd:
    if x in "abcdefghijklmnopqrstuvwxyz":
        has_lower = True

if has_lower == True:
    score = score + 1
else:
    msg.append("Add at least one lowercase letter.")

has_num = False
for c in pwd:
    if c in "0123456789":
        has_num = True

if has_num == True:
    score = score + 1
else:
    msg.append("Add at least one number.")

specials = "!@#$%^&*()-_=+[]{};:'\",.<>?/\\|`~"
got_spec = False
for c in pwd:
    if c in specials:
        got_spec = True

if got_spec == True:
    score = score + 1
else:
    msg.append("Add at least one special character.")

bad_ones = ["password", "123456", "12345678", "qwerty", "admin", "letmein", "welcome"]

if pwd.lower() in bad_ones:
    score = 0
    msg.append("This is a very common password.")

if score >= 6:
    stren = "Very Strong"
elif score >= 4:
    stren = "Strong"
elif score >= 2:
    stren = "Moderate"
else:
    stren = "Weak"

print("\n--------------------------------------------------")
print("Password Analysis")
print("--------------------------------------------------")

print("Length:" , len(pwd))
print("Score:" , score , "/ 6")
print("Strength: " + stren)

if len(msg) > 0:
    print("\nSuggestions:")
    for item in msg:
        print("- " + item)
else:
    print("\nExcellent! Your password meets all the basic checks.")

print("\n" + "="*50)
```
