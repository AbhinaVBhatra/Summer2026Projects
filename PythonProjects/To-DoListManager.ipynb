```python
import json
import os
import sys

fname = "tasks.json"
if os.path.exists(fname):
    f = open(fname, "r")
    tasks = json.load(f)
    f.close()
else:
    tasks = []

while True:
    print("\n==============================")
    print("        TO-DO LIST")
    print("==============================")
    print("1. Add task")
    print("2. View tasks")
    print("3. Complete task")
    print("4. Delete task")
    print("5. Exit")
    print("==============================")

    ch = input("Enter your choice: ")

    if ch == "1":
        title = input("\nEnter the task: ")
        if title == "" or title == " ":
            print("Task cannot be empty.")
        else:
            tasks.append({"title": title, "completed": False})
            f = open(fname, "w")
            json.dump(tasks, f)
            f.close()
            print("Task added successfully!")

    elif ch == "2":
        if len(tasks) == 0:
            print("\nYou don't have any tasks yet.")
        else:
            print("\n---------- YOUR TASKS ----------")
            i = 1
            for t in tasks:
                if t["completed"] == True:
                    status = "✓"
                else:
                    status = " "
                print(i, ". [", status, "]", t["title"])
                i = i + 1
            print("--------------------------------")

    elif ch == "3":
        if len(tasks) == 0:
            print("\nThere are no tasks to complete.")
        else:
            print("\n---------- YOUR TASKS ----------")
            i = 1
            for t in tasks:
                if t["completed"] == True:
                    status = "✓"
                else:
                    status = " "
                print(i, ". [", status, "]", t["title"])
                i = i + 1
            print("--------------------------------")

            num = int(input("Enter the task number to complete: ")) - 1
            if num >= 0 and num < len(tasks):
                tasks[num]["completed"] = True
                f = open(fname, "w")
                json.dump(tasks, f)
                f.close()
                print("Task marked as completed!")
            else:
                print("Invalid task number.")

    elif ch == "4":
        if len(tasks) == 0:
            print("\nThere are no tasks to delete.")
        else:
            print("\n---------- YOUR TASKS ----------")
            i = 1
            for t in tasks:
                if t["completed"] == True:
                    status = "✓"
                else:
                    status = " "
                print(i, ". [", status, "]", t["title"])
                i = i + 1
            print("--------------------------------")

            num = int(input("Enter the task number to delete: ")) - 1
            if num >= 0 and num < len(tasks):
                del_item = tasks.pop(num)
                f = open(fname, "w")
                json.dump(tasks, f)
                f.close()
                print("Deleted:" , del_item['title'])
            else:
                print("Invalid task number.")

    elif ch == "5":
        f = open(fname, "w")
        json.dump(tasks, f)
        f.close()
        print("\nTasks saved. Goodbye!")
        break

    else:
        print("Invalid choice. Please try again.")
```
